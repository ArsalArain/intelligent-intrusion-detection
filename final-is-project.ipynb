```python
# ============================================
# 1. Import required libraries
# ============================================
import numpy as np
import pandas as pd

# Saving / loading (if needed later)
import pickle
from os import path

# Sklearn preprocessing
from sklearn import preprocessing
from sklearn.preprocessing import (
    StandardScaler, OrdinalEncoder, LabelEncoder, MinMaxScaler,
    OneHotEncoder, Normalizer, MaxAbsScaler, RobustScaler, PowerTransformer
)

# Plotting
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn import metrics
from sklearn.metrics import (
    accuracy_score, classification_report, precision_score,
    recall_score, f1_score, roc_auc_score, roc_curve, auc,
    confusion_matrix
)
from sklearn.model_selection import train_test_split

# Deep learning
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from keras.layers import Dense, Conv1D, MaxPool1D, Flatten, Dropout
from keras.models import Sequential

import itertools
from itertools import cycle

# ============================================
# 2. Features & data loading
# ============================================
features = [
    "duration", "protocol_type", "service", "flag", "src_bytes", "dst_bytes",
    "land", "wrong_fragment", "urgent", "hot", "num_failed_logins", "logged_in",
    "num_compromised", "root_shell", "su_attempted", "num_root",
    "num_file_creations", "num_shells", "num_access_files", "num_outbound_cmds",
    "is_host_login", "is_guest_login", "count", "srv_count", "serror_rate",
    "srv_serror_rate", "rerror_rate", "srv_rerror_rate", "same_srv_rate",
    "diff_srv_rate", "srv_diff_host_rate", "dst_host_count",
    "dst_host_srv_count", "dst_host_same_srv_rate", "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate", "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate", "dst_host_srv_serror_rate", "dst_host_rerror_rate",
    "dst_host_srv_rerror_rate", "label", "difficulty"
]

train_path = "https://raw.githubusercontent.com/mohammedAcheddad/AI-Based-Network-IDS_ML-DL/master/nsl-kdd/KDDTrain+.txt"
test_path = "https://raw.githubusercontent.com/mohammedAcheddad/AI-Based-Network-IDS_ML-DL/master/nsl-kdd/KDDTest+.txt"
test21_path = "https://raw.githubusercontent.com/mohammedAcheddad/AI-Based-Network-IDS_ML-DL/master/nsl-kdd/KDDTest-21.txt"  # optional extra test

train_data = pd.read_csv(train_path, names=features)
test_data = pd.read_csv(test_path, names=features)
test21_data = pd.read_csv(test21_path, names=features)

# Drop difficulty (not needed)
for df in (train_data, test_data, test21_data):
    df.drop(columns=['difficulty'], inplace=True)

print("Train shape:", train_data.shape)
print("Test shape:", test_data.shape)
print("Test21 shape:", test21_data.shape)

# ============================================
# 3. Label mapping to 5 classes
# ============================================
def change_label(df):
    df.label.replace(
        ['apache2', 'back', 'land', 'neptune', 'mailbomb', 'pod',
         'processtable', 'smurf', 'teardrop', 'udpstorm', 'worm'],
        'Dos', inplace=True
    )
    df.label.replace(
        ['ftp_write', 'guess_passwd', 'httptunnel', 'imap', 'multihop',
         'named', 'phf', 'sendmail', 'snmpgetattack', 'snmpguess', 'spy',
         'warezclient', 'warezmaster', 'xlock', 'xsnoop'],
        'R2L', inplace=True
    )
    df.label.replace(
        ['ipsweep', 'mscan', 'nmap', 'portsweep', 'saint', 'satan'],
        'Probe', inplace=True
    )
    df.label.replace(
        ['buffer_overflow', 'loadmodule', 'perl', 'ps',
         'rootkit', 'sqlattack', 'xterm'],
        'U2R', inplace=True
    )

change_label(train_data)
change_label(test_data)
change_label(test21_data)

print("\nTrain label distribution:")
print(train_data['label'].value_counts())
print("\nTest label distribution:")
print(test_data['label'].value_counts())

# ============================================
# 4. Optional pie chart for protocol_type and label
# ============================================
def pie_plot(df, cols_list, rows, cols):
    fig, axes = plt.subplots(rows, cols, figsize=(15, 15))
    axes = np.array(axes).reshape(-1)  # flatten in case of 1D
    for ax, col in zip(axes, cols_list):
        df[col].value_counts().plot(
            ax=ax, kind='pie', fontsize=10, autopct='%1.0f%%'
        )
        ax.set_title(str(col), fontsize=12)
    plt.tight_layout()
    plt.show()

pie_plot(train_data, ['protocol_type', 'label'], 1, 2)

# Make working copies
multi_data = train_data.copy()
multi_data_test = test_data.copy()

# ============================================
# 5. Standardization of numeric columns
# ============================================
numeric_cols_train = multi_data.select_dtypes(include='number').columns
numeric_cols_test = multi_data_test.select_dtypes(include='number').columns

# We only want to scale *feature* numeric columns, not the label
numeric_cols_train = [c for c in numeric_cols_train if c != 'difficulty' and c != 'label']
numeric_cols_test = [c for c in numeric_cols_test if c != 'difficulty' and c != 'label']

std_scaler = StandardScaler()

multi_data[numeric_cols_train] = std_scaler.fit_transform(multi_data[numeric_cols_train])
multi_data_test[numeric_cols_test] = std_scaler.transform(multi_data_test[numeric_cols_test])

# ============================================
# 6. Label encoding to integer (0..4) for 5 classes
#    (Dos, normal, Probe, R2L, U2R) – order given by encoder
# ============================================
le = preprocessing.LabelEncoder()
multi_data['intrusion'] = le.fit_transform(multi_data['label'])
multi_data_test['intrusion'] = le.transform(multi_data_test['label'])

print("\nEncoded classes (LabelEncoder):", list(le.classes_))

# Drop original label
multi_data.drop(columns=['label'], inplace=True)
multi_data_test.drop(columns=['label'], inplace=True)

# ============================================
# 7. One-hot encoding of categorical features
# ============================================
cat_cols = ['protocol_type', 'service', 'flag']

X_train_df = multi_data.drop(columns=['intrusion'])
X_test_df = multi_data_test.drop(columns=['intrusion'])

X_train_df = pd.get_dummies(X_train_df, columns=cat_cols, prefix="", prefix_sep="")
X_test_df = pd.get_dummies(X_test_df, columns=cat_cols, prefix="", prefix_sep="")

# Align columns (same feature space for train & test)
X_train_df, X_test_df = X_train_df.align(X_test_df, join='inner', axis=1, fill_value=0)

y_train_int = multi_data['intrusion'].values
y_test_int = multi_data_test['intrusion'].values

print('\nX_train shape:', X_train_df.shape)
print('X_test shape:', X_test_df.shape)

# ============================================
# 8. Prepare data for CNN (reshape + one-hot labels)
# ============================================
from sklearn.preprocessing import LabelBinarizer

num_classes = len(le.classes_)

# One-hot encoded y for CNN / ROC curves
lb = LabelBinarizer()
lb.fit(y_train_int)
y_train_multi = lb.transform(y_train_int)
y_test_multi = lb.transform(y_test_int)

# In case LabelBinarizer returns shape (n,) for binary – ensure 2D
if y_train_multi.ndim == 1:
    y_train_multi = np.vstack((1 - y_train_multi, y_train_multi)).T
    y_test_multi = np.vstack((1 - y_test_multi, y_test_multi)).T

X_train = X_train_df.values.astype('float32')
X_test = X_test_df.values.astype('float32')

X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

print("\nCNN input shapes:")
print("X_train:", X_train.shape, "y_train:", y_train_multi.shape)
print("X_test:", X_test.shape, "y_test:", y_test_multi.shape)

# ============================================
# 9. Build CNN model
# ============================================
model = Sequential()
model.add(
    Conv1D(
        32, 3, padding="same",
        input_shape=(X_train.shape[1], 1),
        activation='relu'
    )
)
model.add(MaxPool1D(pool_size=4))
model.add(Dropout(0.2))

model.add(Conv1D(32, 3, padding="same", activation='relu'))
model.add(MaxPool1D(pool_size=4))
model.add(Dropout(0.2))

model.add(Flatten())
model.add(Dense(units=50, activation='relu'))

# Output layer: number of classes
model.add(Dense(units=y_train_multi.shape[1], activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

# ============================================
# 10. Train CNN model (GPU if available)
# ============================================
device = '/GPU:0' if tf.config.list_physical_devices('GPU') else '/CPU:0'
print("\nTraining on device:", device)

with tf.device(device):
    history = model.fit(
        X_train, y_train_multi,
        epochs=100,
        batch_size=5000,
        validation_split=0.2,
        verbose=1
    )

# Check dtypes
print("\nX_train dtype:", X_train.dtype)
print("X_test dtype:", X_test.dtype)

# ============================================
# 11. Evaluate CNN
# ============================================
test_results = model.evaluate(X_test, y_test_multi, verbose=1)
print(f'\nTest results - Loss: {test_results[0]:.4f} - Accuracy: {test_results[1] * 100:.2f}%')

# Plot accuracy vs epoch
plt.figure()
plt.plot(history.history['accuracy'], label='Train accuracy')
plt.plot(history.history['val_accuracy'], label='Validation accuracy')
plt.title("Accuracy vs Epoch")
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend()
plt.show()

# ============================================
# 12. Predictions, AUC, ROC, Confusion Matrix
# ============================================
y_pred_proba = model.predict(X_test, batch_size=500)

# For ROC curve plotting (one-hot like)
y_pred_onehot = to_categorical(np.argmax(y_pred_proba, axis=1), y_train_multi.shape[1])
y_test_onehot = y_test_multi

y_pred_argmax = np.argmax(y_pred_proba, axis=1)
y_test_argmax = np.argmax(y_test_onehot, axis=1)

# ---------- AUC per-class ----------
def AUC(actual_class, pred_class, average="micro"):
    unique_class = set(actual_class)
    roc_auc_dict = {}
    for per_class in unique_class:
        other_class = [x for x in unique_class if x != per_class]
        new_actual_class = [0 if x in other_class else 1 for x in actual_class]
        new_pred_class = [0 if x in other_class else 1 for x in pred_class]
        roc_auc = roc_auc_score(new_actual_class, new_pred_class, average=average)
        roc_auc_dict[per_class] = roc_auc
    return roc_auc_dict

print('\nAUC Score per class (on Test):', AUC(y_test_argmax, y_pred_argmax))

# ---------- ROC Curve plotting ----------
def plot_roc_curve(y_test, y_pred, classes):
    n_classes = y_test.shape[1]
    fpr = dict()
    tpr = dict()
    roc_auc = dict()

    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_test[:, i], y_pred[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])

    fpr["micro"], tpr["micro"], _ = roc_curve(y_test.ravel(), y_pred.ravel())
    roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

    all_fpr = np.unique(np.concatenate([fpr[i] for i in range(n_classes)]))
    mean_tpr = np.zeros_like(all_fpr)
    for i in range(n_classes):
        mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
    mean_tpr /= n_classes

    fpr["macro"] = all_fpr
    tpr["macro"] = mean_tpr
    roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])

    lw = 2
    plt.figure(figsize=(10, 10))
    plt.plot(
        fpr["micro"], tpr["micro"],
        label=f"micro-average ROC curve (area = {roc_auc['micro']:.2f})",
        linestyle=":", linewidth=4
    )
    plt.plot(
        fpr["macro"], tpr["macro"],
        label=f"macro-average ROC curve (area = {roc_auc['macro']:.2f})",
        linestyle=":", linewidth=4
    )

    colors = cycle(["aqua", "darkorange", "cornflowerblue", "green", "red"])
    for i, color in zip(range(n_classes), colors):
        plt.plot(
            fpr[i], tpr[i],
            color=color, lw=lw,
            label=f"ROC curve of {classes[i]} (area = {roc_auc[i]:0.2f})"
        )

    plt.plot([0, 1], [0, 1], "k--", lw=lw)
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("Multi-class ROC Curves")
    plt.legend(loc="lower right")
    plt.show()

classes_5 = ['normal', 'Dos', 'Probe', 'R2L', 'U2R']  # desired order
plot_roc_curve(y_test_onehot, y_pred_onehot, classes_5)

# ---------- Confusion Matrix (DL) ----------
def plot_confusion_matrix(cm, classes, normalize=False,
                          title='Confusion matrix',
                          cmap=plt.cm.Blues):
    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        print("\nNormalized confusion matrix")
    else:
        print('\nConfusion matrix, without normalization')

    plt.imshow(cm, interpolation='nearest', cmap=cmap)
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)

    fmt = '.2f' if normalize else 'd'
    thresh = cm.max() / 2.
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(
            j, i, format(cm[i, j], fmt),
            horizontalalignment="center",
            color="white" if cm[i, j] > thresh else "black"
        )

    plt.tight_layout()
    plt.ylabel('True label')
    plt.xlabel('Predicted label')

cm_dl = confusion_matrix(y_test_argmax, y_pred_argmax)
plt.figure(figsize=(7, 7))
plot_confusion_matrix(cm_dl, classes=classes_5, title='DL Confusion Matrix')
plt.show()

# ============================================
# 13. RandomForest classifier on same features
# ============================================
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train_df, y_train_int)
rf_pred = rf.predict(X_test_df)

print("\nRandomForest Classification Report (Test):")
print(classification_report(y_test_int, rf_pred, target_names=le.classes_))

cm_rf = confusion_matrix(y_test_int, rf_pred)
plt.figure(figsize=(7, 7))
plot_confusion_matrix(cm_rf, classes=le.classes_, title='RF Confusion Matrix')
plt.show()
```

```python
# ============================================================
# NSL-KDD MULTI-CLASS: TRAIN CNN + RF, THEN SAVE ARTIFACTS
# ============================================================
import os, json
from pathlib import Path

import numpy as np
import pandas as pd

# Plotting (optional)
import matplotlib.pyplot as plt

# Sklearn
from sklearn import preprocessing
from sklearn.preprocessing import StandardScaler, LabelBinarizer
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, auc
from sklearn.ensemble import RandomForestClassifier

# Deep learning
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from keras.layers import Dense, Conv1D, MaxPool1D, Flatten, Dropout
from keras.models import Sequential
import joblib

# -----------------------------
# Paths / Artifacts dir
# -----------------------------
ARTIFACT_DIR = Path("/kaggle/working") if Path("/kaggle/working").exists() else Path(".")
ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)

# -----------------------------
# 1) Load data
# -----------------------------
features = [
    "duration","protocol_type","service","flag","src_bytes","dst_bytes","land",
    "wrong_fragment","urgent","hot","num_failed_logins","logged_in","num_compromised",
    "root_shell","su_attempted","num_root","num_file_creations","num_shells",
    "num_access_files","num_outbound_cmds","is_host_login","is_guest_login","count",
    "srv_count","serror_rate","srv_serror_rate","rerror_rate","srv_rerror_rate",
    "same_srv_rate","diff_srv_rate","srv_diff_host_rate","dst_host_count",
    "dst_host_srv_count","dst_host_same_srv_rate","dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate","dst_host_srv_diff_host_rate","dst_host_serror_rate",
    "dst_host_srv_serror_rate","dst_host_rerror_rate","dst_host_srv_rerror_rate",
    "label","difficulty"
]

train_url = "https://raw.githubusercontent.com/mohammedAcheddad/AI-Based-Network-IDS_ML-DL/master/nsl-kdd/KDDTrain+.txt"
test_url  = "https://raw.githubusercontent.com/mohammedAcheddad/AI-Based-Network-IDS_ML-DL/master/nsl-kdd/KDDTest+.txt"
test21_url= "https://raw.githubusercontent.com/mohammedAcheddad/AI-Based-Network-IDS_ML-DL/master/nsl-kdd/KDDTest-21.txt"

train_df = pd.read_csv(train_url, names=features)
test_df  = pd.read_csv(test_url,  names=features)
test21_df= pd.read_csv(test21_url, names=features)

for df in (train_df, test_df, test21_df):
    df.drop(columns=["difficulty"], inplace=True)

# -----------------------------
# 2) Collapse labels into 5 families
# -----------------------------
def change_label(df):
    df.label.replace(['apache2','back','land','neptune','mailbomb','pod','processtable',
                      'smurf','teardrop','udpstorm','worm'],'Dos', inplace=True)
    df.label.replace(['ftp_write','guess_passwd','httptunnel','imap','multihop','named','phf',
                      'sendmail','snmpgetattack','snmpguess','spy','warezclient','warezmaster',
                      'xlock','xsnoop'],'R2L', inplace=True)      
    df.label.replace(['ipsweep','mscan','nmap','portsweep','saint','satan'],'Probe', inplace=True)
    df.label.replace(['buffer_overflow','loadmodule','perl','ps','rootkit','sqlattack','xterm'],'U2R', inplace=True)

for df in (train_df, test_df, test21_df):
    change_label(df)

# Working copies
train = train_df.copy()
test  = test_df.copy()

# -----------------------------
# 3) Scale numeric columns
# -----------------------------
numeric_cols_train = [c for c in train.select_dtypes(include='number').columns if c != 'label']
numeric_cols_test  = [c for c in test.select_dtypes(include='number').columns if c != 'label']

scaler = StandardScaler()
train[numeric_cols_train] = scaler.fit_transform(train[numeric_cols_train])
test[numeric_cols_test]   = scaler.transform(test[numeric_cols_test])

# -----------------------------
# 4) Encode label -> integers (0..4)
# -----------------------------
le = preprocessing.LabelEncoder()
train['intrusion'] = le.fit_transform(train['label'])
test['intrusion']  = le.transform(test['label'])

# Keep a readable order for later
class_names = list(le.classes_)
print("Classes:", class_names)

# -----------------------------
# 5) One-hot encode categoricals via get_dummies
#    (We must save the final column order!)
# -----------------------------
cat_cols = ['protocol_type','service','flag']

X_train_df = pd.get_dummies(
    train.drop(columns=['label','intrusion']),
    columns=cat_cols, prefix="", prefix_sep=""
)
X_test_df = pd.get_dummies(
    test.drop(columns=['label','intrusion']),
    columns=cat_cols, prefix="", prefix_sep=""
)

# Align
X_train_df, X_test_df = X_train_df.align(X_test_df, join='inner', axis=1, fill_value=0)

y_train_int = train['intrusion'].values
y_test_int  = test['intrusion'].values

# Save the final feature order for inference
onehot_columns = list(X_train_df.columns)
with open(ARTIFACT_DIR / "onehot_columns.json", "w") as f:
    json.dump(onehot_columns, f, indent=2)

# -----------------------------
# 6) CNN data shapes
# -----------------------------
# Make one-hot Y for NN / ROC curves
lb = LabelBinarizer()
lb.fit(y_train_int)
y_train_multi = lb.transform(y_train_int)
y_test_multi  = lb.transform(y_test_int)

if y_train_multi.ndim == 1:
    y_train_multi = np.vstack((1 - y_train_multi, y_train_multi)).T
    y_test_multi  = np.vstack((1 - y_test_multi, y_test_multi)).T

X_train = X_train_df.values.astype('float32')
X_test  = X_test_df.values.astype('float32')
X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
X_test  = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

# -----------------------------
# 7) Build & train CNN
# -----------------------------
model = Sequential([
    Conv1D(32, 3, padding="same", activation='relu', input_shape=(X_train.shape[1], 1)),
    MaxPool1D(pool_size=4),
    Dropout(0.2),
    Conv1D(32, 3, padding="same", activation='relu'),
    MaxPool1D(pool_size=4),
    Dropout(0.2),
    Flatten(),
    Dense(50, activation='relu'),
    Dense(y_train_multi.shape[1], activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
device = '/GPU:0' if tf.config.list_physical_devices('GPU') else '/CPU:0'
print("Training device:", device)

with tf.device(device):
    history = model.fit(
        X_train, y_train_multi, epochs=100, batch_size=5000,
        validation_split=0.2, verbose=1
    )

# -----------------------------
# 8) Evaluate CNN
# -----------------------------
test_loss, test_acc = model.evaluate(X_test, y_test_multi, verbose=0)
print(f"CNN Test: loss={test_loss:.4f}, acc={test_acc*100:.2f}%")

# -----------------------------
# 9) Train RandomForest on same features
# -----------------------------
rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_train_df, y_train_int)
rf_pred = rf.predict(X_test_df)

print("\nRandomForest report:")
print(classification_report(y_test_int, rf_pred, target_names=class_names))

# -----------------------------
# 10) SAVE ARTIFACTS
# -----------------------------
# CNN
cnn_path = ARTIFACT_DIR / "cnn_model.h5"
model.save(cnn_path)

# RF
rf_path = ARTIFACT_DIR / "rf_model.pkl"
joblib.dump(rf, rf_path)

# Scaler for numeric cols
scaler_path = ARTIFACT_DIR / "scaler.pkl"
joblib.dump(scaler, scaler_path)

# LabelEncoder
le_path = ARTIFACT_DIR / "label_encoder.pkl"
joblib.dump(le, le_path)

# Also save which numeric columns were scaled (safe for inference)
with open(ARTIFACT_DIR / "numeric_columns.json", "w") as f:
    json.dump(sorted(numeric_cols_train), f, indent=2)

print("\nSaved artifacts to:", ARTIFACT_DIR.resolve())
print(" -", cnn_path.name)
print(" -", rf_path.name)
print(" -", scaler_path.name)
print(" -", le_path.name)
print(" - onehot_columns.json")
print(" - numeric_columns.json")

# -----------------------------
# 11) (Optional) Quick loaders to sanity-check artifacts
# -----------------------------
def _load_artifacts_check():
    import tensorflow as tf, joblib, json
    from pathlib import Path
    base = ARTIFACT_DIR
    _cnn = tf.keras.models.load_model(base / "cnn_model.h5")
    _rf  = joblib.load(base / "rf_model.pkl")
    _sc  = joblib.load(base / "scaler.pkl")
    _le  = joblib.load(base / "label_encoder.pkl")
    with open(base / "onehot_columns.json") as f:
        _cols = json.load(f)
    with open(base / "numeric_columns.json") as f:
        _numc = json.load(f)
    print("Loaded OK. Classes:", list(_le.classes_))
    return True

_ = _load_artifacts_check()
```
